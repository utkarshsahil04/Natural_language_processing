### Q1 - group words by stem

change the words in the list and run the cell

```python
from nltk.stem import PorterStemmer

stem = PorterStemmer()

# put your words here (space separated string also fine)
text = "running runner runs easily easier easy cats cat happier happiness"
words = text.lower().split()

families = {}
for word in words:
    root = stem.stem(word)
    if root in families:
        families[root].append(word)
    else:
        families[root] = [word]

print("Morphological Word Families:\n")
for root, group in sorted(families.items()):
    print(f"{root} ({len(group)}): {group}")
```

### Q2 - check prefix suffix on a word

longer prefixes/suffixes are checked first so `pre` wins over `p`, `tion` over shorter endings, etc.

```python
from nltk.stem import PorterStemmer

stem = PorterStemmer()

prefixes = ["un", "re", "dis", "mis", "pre", "non", "in", "im"]
suffixes = ["ing", "ed", "ly", "ness", "able", "ful", "less", "ment", "tion"]

# longer ones first (important)
prefixes = sorted(prefixes, key=len, reverse=True)
suffixes = sorted(suffixes, key=len, reverse=True)

# change this word and run
word = "unhappiness"
word = word.lower().strip()

found_prefix = None
found_suffix = None

for p in prefixes:
    # avoid matching tiny prefix on very short words
    if word.startswith(p) and len(word) > len(p) + 1:
        found_prefix = p
        break

for s in suffixes:
    if word.endswith(s) and len(word) > len(s) + 1:
        found_suffix = s
        break

if found_prefix and found_suffix:
    result = f"both prefix ('{found_prefix}') and suffix ('{found_suffix}')"
elif found_prefix:
    result = f"prefix ('{found_prefix}')"
elif found_suffix:
    result = f"suffix ('{found_suffix}')"
else:
    result = "neither"

print("word          :", word)
print("stem          :", stem.stem(word))
print("classification:", result)
```

### try more words for Q2

```python
from nltk.stem import PorterStemmer

stem = PorterStemmer()
prefixes = sorted(["un", "re", "dis", "mis", "pre", "non", "in", "im"], key=len, reverse=True)
suffixes = sorted(["ing", "ed", "ly", "ness", "able", "ful", "less", "ment", "tion"], key=len, reverse=True)

test_words = ["running", "undo", "helpful", "careless", "preview", "mismatch", "nation", "happily", "book"]

for word in test_words:
    word = word.lower()
    found_prefix = None
    found_suffix = None

    for p in prefixes:
        if word.startswith(p) and len(word) > len(p) + 1:
            found_prefix = p
            break

    for s in suffixes:
        if word.endswith(s) and len(word) > len(s) + 1:
            found_suffix = s
            break

    if found_prefix and found_suffix:
        result = f"both ('{found_prefix}' + '{found_suffix}')"
    elif found_prefix:
        result = f"prefix ('{found_prefix}')"
    elif found_suffix:
        result = f"suffix ('{found_suffix}')"
    else:
        result = "neither"

    print(f"{word:12} | stem={stem.stem(word):8} | {result}")
```
