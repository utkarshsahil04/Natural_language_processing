### Q1 - group words by stem

```python
from nltk.stem import PorterStemmer

stem = PorterStemmer()
words = input("Enter words separated by spaces: ").lower().split()

families = {}
for word in words:
    root = stem.stem(word)
    if root in families:
        families[root].append(word)
    else:
        families[root] = [word]

print("\nMorphological Word Families:\n")
for root, group in sorted(families.items()):
    print(f"{root} ({len(group)}): {group}")
```

### Q2 - check prefix suffix on a word

```python
from nltk.stem import PorterStemmer

stem = PorterStemmer()
prefixes = ["un", "re", "dis", "mis", "pre", "non", "in", "im"]
suffixes = ["ing", "ed", "ly", "ness", "able", "ful", "less", "ment", "tion"]

word = input("enter word: ").lower()
found_prefix = None
found_suffix = None

for p in prefixes:
    if word.startswith(p):
        found_prefix = p
        break

for s in suffixes:
    if word.endswith(s):
        found_suffix = s
        break

if found_prefix and found_suffix:
    result = f"both prefix ('{found_prefix}') and suffix ('{found_suffix}')"
elif found_prefix:
    result = f"prefix ('{found_prefix}')"
elif found_suffix:
    result = f"suffix ('{found_suffix}')"
else:
    result = "neither"

print("word          :", word)
print("stem          :", stem.stem(word))
print("classification:", result)
```
