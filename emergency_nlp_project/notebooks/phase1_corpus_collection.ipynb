# Phase 1 - Corpus Collection & Exploration

**Dataset:** NLP with Disaster Tweets (Kaggle)
**Goal:** Build a clean CSV of 300-500+ real emergency messages


## Step 0 - Kaggle Setup Instructions

1. Go to https://www.kaggle.com/settings -> API -> **Create New Token**
2. A file `kaggle.json` will download. Place it at:
   - Windows: `C:/Users/<YourName>/.kaggle/kaggle.json`
3. Run the download script below OR run the cell directly


```python
import os, sys
# Add project root to path
project_root = os.path.abspath(os.path.join(os.getcwd(), '..'))
sys.path.insert(0, project_root)
print('Project root:', project_root)
```

```python
# Download data using Kaggle API
# Make sure kaggle.json is in ~/.kaggle/
RAW_DIR = os.path.join(project_root, 'data', 'raw')
os.makedirs(RAW_DIR, exist_ok=True)

if not os.path.exists(os.path.join(RAW_DIR, 'train.csv')):
    os.system(f'kaggle competitions download -c nlp-getting-started -p "{RAW_DIR}"')
    import zipfile
    zip_path = os.path.join(RAW_DIR, 'nlp-getting-started.zip')
    with zipfile.ZipFile(zip_path, 'r') as z:
        z.extractall(RAW_DIR)
    print('Downloaded and extracted!')
else:
    print('Data already downloaded.')
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('seaborn-v0_8-darkgrid')

df = pd.read_csv(os.path.join(RAW_DIR, 'train.csv'))
print('Shape:', df.shape)
df.head(10)
```

## Step 1 - Dataset Overview

```python
print('=== Column Info ===')
df.info()
print()
print('=== Target Distribution ===')
print(df['target'].value_counts())
print()
print('=== Missing Values ===')
print(df.isnull().sum())
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Class distribution
df['target'].value_counts().plot(kind='bar', ax=axes[0], color=['steelblue','tomato'])
axes[0].set_title('Disaster vs Non-Disaster Tweets')
axes[0].set_xticklabels(['Non-Disaster (0)', 'Disaster (1)'], rotation=0)
axes[0].set_ylabel('Count')

# Message length distribution
df['text_length'] = df['text'].str.len()
df.groupby('target')['text_length'].plot(kind='hist', bins=30, alpha=0.6, ax=axes[1])
axes[1].set_title('Message Length Distribution')
axes[1].set_xlabel('Character Count')
axes[1].legend(['Non-Disaster', 'Disaster'])

plt.tight_layout()
plt.savefig(os.path.join(project_root, 'reports', 'phase1_eda.png'), dpi=150)
plt.show()
```

## Step 2 - Filter Disaster Messages Only

```python
# Filter to real disaster messages only
disaster_df = df[df['target'] == 1][['id','text','keyword','location']].copy()
disaster_df = disaster_df.rename(columns={'id':'message_id','text':'raw_text'})
disaster_df['source'] = 'kaggle_nlp_disaster_tweets'
disaster_df['message_id'] = disaster_df['message_id'].apply(lambda x: f'KG_{int(x):05d}')
print(f'Disaster messages: {len(disaster_df)}')
disaster_df.head()
```

## Step 4 - Merge & Clean Final Corpus

```python
corpus = pd.concat(
    [disaster_df[['message_id','raw_text','source']], hinglish_df],
    ignore_index=True
)

# Clean
corpus['raw_text'] = corpus['raw_text'].str.strip()
corpus = corpus[corpus['raw_text'].str.len() > 10]
corpus = corpus.drop_duplicates(subset=['raw_text'])
corpus = corpus.reset_index(drop=True)

print(f'Final corpus size: {len(corpus)}')
corpus.head(10)
```

```python
# Source breakdown
print('Source breakdown:')
print(corpus['source'].value_counts())
print()
print('Message length stats:')
corpus['text_length'] = corpus['raw_text'].str.len()
print(corpus['text_length'].describe())
```

```python
# Save corpus
PROC_DIR = os.path.join(project_root, 'data', 'processed')
os.makedirs(PROC_DIR, exist_ok=True)
out_path = os.path.join(PROC_DIR, 'corpus.csv')
corpus[['message_id','raw_text','source']].to_csv(out_path, index=False, encoding='utf-8')
print(f'Corpus saved to: {out_path}')
print(f'Total messages: {len(corpus)}')
```
